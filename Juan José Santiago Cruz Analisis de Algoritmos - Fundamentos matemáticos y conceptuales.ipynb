- *¿Qué establece el Principio de Invarianza respecto al tiempo de ejecución de dos implementaciones distintas para un mismo algoritmo?*  
**RESPUESTA:**  
El Principio de Invarianza establece que el tiempo de ejecución de dos implementaciones distintas de un mismo algoritmo no diferirá más que en una constante multiplicativa1more_horiz. Esto se aplica tanto si las implementaciones se ejecutan en máquinas distintas como si se escriben con códigos o lenguajes diferentes.  
**Definición Formal **
Si se tiene un algoritmo y dos implementaciones suyas, $I_1$ e $I_2$, cuyos tiempos de ejecución son $T_1(n)$ y $T_2(n)$ respectivamente, el principio afirma que existe una constante real $c > 0$ y un número natural $n_0$ tales que para todo tamaño de entrada $n \geq n_0$ se verifica que1more_horiz:$$T_1(n) \leq c \cdot T_2(n)$$  
**Implicaciones Principales**  
**Dependencia del algoritmo, no del entorno:** El tiempo requerido para resolver un problema mediante un algoritmo depende de la naturaleza intrínseca del propio algoritmo y no de factores externos de la implementación como el procesador, el compilador o el lenguaje de programación elegido2more_horiz.  
**Mismo comportamiento asintótico:** Si una de las implementaciones pertenece a un determinado orden de complejidad (por ejemplo, una implementación $I_1$ que es de orden $O(n^2)$), bajo el Principio de Invarianza se garantiza que la otra implementación $I_2$ también será del mismo orden de complejidad ($O(n^2)$).

**RESPUESTA:**  
1. **¿Qué es un Algoritmo?** (La "receta")  
Un algoritmo es simplemente una receta paso a paso para resolver un problema. Para que sea un algoritmo de verdad, debe cumplir tres reglas de oro:  
**Tiene que terminar:** No puede durar para siempre (tiene un final obligatorio).  
**No debe tener dudas:** Cada paso es exacto y claro (una computadora o persona puede seguirlo al pie de la letra sin adivinar).  
**Es consistente:** Si le das los mismos datos de inicio, siempre te dará el mismo resultado.
2. **El Problema del Paro** (La pantalla congelada)  
Imagina que quieres inventar un programa "mágico" que sea capaz de revisar cualquier otro código y decirte con 100% de seguridad: "Oye, este programa sí va a terminar" o "Cuidado, este programa se va a congelar para siempre".  
**Lo que demostró Alan Turing:** Descubrió que ese programa "mágico" es matemáticamente imposible de crear. No existe un método general que funcione para todos los casos; siempre habrá algún código que confunda a tu analizador.
3. **"En papel"** (A Priori) vs. **"Con cronómetro"** (A Posteriori)  
Para medir qué tan rápido es tu algoritmo, tienes dos caminos:  
**Análisis a Priori (Teórico):** Calculas la eficiencia con matemáticas en el papel, antes de programar. Analizas el diseño y los pasos del algoritmo. No depende de qué computadora uses, por lo que es una medida universal.  
**Prueba a Posteriori (Práctico):** Primero escribes el código, lo corres en una computadora real y mides con un reloj cuántos segundos tarda. Depende totalmente de tu máquina (si tienes una computadora vieja, tardará más que en una nueva).

1. Implementa una versión recursiva para el cálculo del factorial ($n! = n \cdot (n-1)!$) y una versión iterativa para Fibonacci.

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n-1)

def fibonacci_iterativo(n):
  if n == 0:
    return 0
  if n == 1:
    return 1
  anterior = 0;
  actual = 1;
  for i in range (2, n + 1):
    nuevo_fibo = actual + anterior;
    anterior = actual;
    actual = nuevo_fibo;
  return actual
print(factorial(5), fibonacci_iterativo(35))
````
